```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
import statistics as st
```

```python
# Data download from sklearn
from sklearn.datasets import load_iris
data=load_iris().data
target=load_iris().target
df_data=pd.DataFrame(data,columns=['sepal_length','sepal_width','petal_length','petal_width'])
df_target=pd.DataFrame(target,columns=['target'])
```

```python
df_merged=pd.merge(df_data, df_target, left_index=True, right_index=True)
```

Nous constatons qu'il n'y a pas de données manquantes, nous pouvons donc procéder à l'analyse.

```python
X=df_merged[['sepal_length','sepal_width','petal_length','petal_width']]
y=df_merged['target']

print(X.shape)
print(y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
%%time
from sklearn.neighbors import KNeighborsClassifier
knn_clf = KNeighborsClassifier(5)
knn_clf.fit(X_train, y_train)
```

```python
y_pred = knn_clf.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test, y_pred))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
cm = confusion_matrix(y_test, y_pred)
disp=ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot()
```

Evaluate impact of accuracy

```python
k=[10,20,30,40,50]

average=[]
correct_predictions=[]

for i in k:
    score=[]
    correct=[]
    for j in range(10):
        
        X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
        knn_clf = KNeighborsClassifier(i)
        knn_clf.fit(X_train, y_train)
        y_pred = knn_clf.predict(X_test)
        score.append(round(accuracy_score(y_test, y_pred),2))    #we'll store the accuracy score of each run in the score list
        
        #accuracy score = number of correct predictions / total number of predictions
        #so number of correct predictions=accuracy score*total number of predictions
        # from data splitting info, we know that total number of predictions is 30
            
        corr=accuracy_score(y_test, y_pred)*30  #here we're calculating the number of correct predictions
        correct.append(round(corr))   #number of correct predictions will be stored in the correct list
        
    average.append(round(st.mean(score),2)) #this list contains the average accuracy score for each k
    correct_predictions.append(round(st.mean(correct))) #this list contains the average number of correct predictions for each k
    
#let's create a dataframe to store all values

kimpact_df=pd.DataFrame(data=[1,3,5,7,10,20,30,40,50], columns=['k_values'])
kimpact_df['average_score']=average
kimpact_df['correct_predictions']=correct_predictions
kimpact_df
```
